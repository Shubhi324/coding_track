```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv('coding_track.csv')

# Convert DATE column to datetime for better plotting
df['DATE'] = pd.to_datetime(df['DATE'], format='%d-%b')

# Plotting
plt.figure(figsize=(10, 6))

# Plot each platform with different color and marker
plt.plot(df['DATE'], df['GFG'], marker='o', color='green', linewidth=2, label='GFG')
plt.plot(df['DATE'], df['A2Z'], marker='s', color='blue', linewidth=2, label='A2Z')
plt.plot(df['DATE'], df['hackerrank'], marker='^', color='orange', linewidth=2, label='Hackerrank')

# Customizing chart
plt.title('Questions Solved Per Platform', fontsize=14, fontweight='bold')
plt.xlabel('Date', fontsize=12)
plt.ylabel('Number of Questions', fontsize=12)
plt.xticks(df['DATE'], df['DATE'].dt.strftime('%d-%b'))  # Format x-axis as 30-Apr etc.
plt.legend(title='Platform')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Show the plot
plt.show()
```
